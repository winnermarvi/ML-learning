```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("data\multipleLinearRegression_scratch.csv")
df.head()
```

```python
X = df.iloc[:,0:4]
y = df.iloc[:,4]
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
class Multiple_linear_regression:

    def __init__(self):
        self.intercept_ = None
        self.coef_ = None

    def fit(self,X_train,y_train):
        X_train = np.insert(X_train,0,1,axis=1)
        betas = np.linalg.inv(X_train.T @ X_train) @ X_train.T @ y_train
        self.coef_ = betas[1:]
        self.intercept_ = betas[0]


    def predict(self,X_test):
        
        y_pred = self.intercept_ + X_test @ self.coef_
        return np.array(y_pred)
```

```python
LR = Multiple_linear_regression()
```

```python
LR.fit(X_train,y_train)
y_pred1 = LR.predict(X_test)
```

```python
sklr = LinearRegression()
sklr.fit(X_train,y_train)
y_pred2 = sklr.predict(X_test)
```
